```python
# which customers spend the most
df.groupby("Customer_ID")["Total_Amount"].sum().sort_values(ascending=False).head(10)
```

```python
# which customer place the most orders
df.groupby("Customer_ID")["Order_ID"].count().sort_values(ascending=False).head(10)
```

```python
# Which product category is purchesed the most
df.groupby("Product_Category")["Quantity"].sum().sort_values(ascending=False)
```

```python
# Which product category generates the higest revenue
df.groupby("Product_Category")["Total_Amount"].sum().sort_values(ascending=False)
```

```python
# What is the average order value
df["Total_Amount"].mean()
```

```python
# How much revenue comes from new vs returning customers
df.groupby("Is_Returning_Customer")["Total_Amount"].sum()
```

```python
# which city generates the higest revenue
df.groupby("City")["Total_Amount"].sum().sort_values(ascending=False).head(10)
```

```python
# which city has the most orders
df.groupby("City")["Order_ID"].count().sort_values(ascending=False).head(10)
```

```python
# which payment method is used the most
df["Payment_Method"].value_counts()
```

```python
# which device generates the highest revenue
df.groupby("Device_Type")["Total_Amount"].sum() .sort_values(ascending=False)
```

```python
# Does discount have a relationship with spending
df[["Discount_Amount","Total_Amount"]].corr()
```

```python
#Does session duration affect spending
df[["Session_Duration_Minutes","Total_Amount"]].corr()
```

```python
# Does viewing more pages relate to higher spending
df[["Pages_Viewed","Total_Amount"]].corr()
```

```python
# does delivery time affect customers rating
df.groupby("Delivery_Time_Days")["Customer_Rating"].mean()
```

```python
# which age group spend most
bins=[0,18,25,35,45,55,100]

lable=["<18","18-25","26-35","36-45","46-55","56+"]

df["Age_Group"]=pd.cut(df["Age"],bins=bins,
labels=lable)

df.groupby("Age_Group",observed=True)["Total_Amount"].sum().sort_values(ascending=False)
                       
```

```python
# which gender has higher average spending
df.groupby("Gender")["Total_Amount"].mean().sort_values(ascending=False)
```

```python
# what is the average quantity purchased per order 
df["Quantity"].mean()
```

```python
# what percentage of customer are returning customers
df["Customer_Rating"].mean()
```
